```python
import tensorflow as tf
```

```python
mnist = tf.keras.datasets.mnist # AMBIL OBJEK DATASET MNIST YG DISEDIAKAN TENSORFLOW
(x_train, y_train), (x_test, y_test) = mnist.load_data() # buat dataset (bagi 2 dtraining & dtesting)
x_train, x_test = x_train / 255.0, x_test / 255.0 # normalisasi gambar pixel (0-255 nilainya)
# neural network lebih mudah b elajar jika semua fitur berada di skala sama
```

```python
# BUAT MODEL (SEQUENTIAL = DATA NGALIR DARI ATAS KE BAWAH)
model = tf.keras.models.Sequential([tf.keras.layers.Flatten(input_shape=(28,28)),
                                    # ubah jadi 28 x 28 = 784 angka
                                    tf.keras.layers.Dense(128, activation='relu'),
                                    # 784 -> 128, tiap neuron trima semua input (Fully Connected Layer)
                                    # ReLU = fungsi -> f(x)=max(0,x)
                                    tf.keras.layers.Dropout(0.2),
                                    # neuron matikan acak ketika training (20% nya) biar ga depending & kurangi overfitting
                                    tf.keras.layers.Dense(10)])
                                    # 10 kelas jadi skor mentah (logits), belum berupa probabilitas
```

```python
predictions = model(x_train[:1]).numpy() # predik 1 gambar pertama trus masukkin ke model
predictions
```

```python
tf.nn.softmax(predictions).numpy() # softmax ubah logits jadi probabilitas
```

```python
loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
# HITUNG LOSS (seberapa jauh prediksi model dari jawaban benar)
loss_fn(y_train[:1], predictions).numpy() #  LOSS NYA GEDE, 0 hampir sempurna; >2 sgt buruk
```

```python
model.compile(optimizer='adam', loss=loss_fn, metrics=['accuracy'])
# compile/konfig proses training
# optimizer pake Adam buat perbarui bobot model agar loss turun
# Loss pake SparseCategoricalCrossentropy sebagai fungsi yang dioptimalkan
# Metrics tampilkan accuracy biar tau prediksi benar
model.fit(x_train, y_train, epochs=5) # epoch: satu siklus penuh di mana seluruh dataset pelatihan (data latih) diproses oleh model.
# prediksi dgn 5 siklus
```

```python
model.evaluate(x_test, y_test, verbose=2)
# evaluasi dengan verbose=2 -> tampilin ringkasan metrik(loss/akurasi dalam bentuk 1 baris teks baru tiap kali 1 epoch selesai, tanpa bilah anomasi)
```

```python
probability_model = tf.keras.Sequential([model, tf.keras.layers.Softmax()])
# MODEL SEBELUMNYA HASILKAN LOGITS (SKOR MENTAH),
# DGN TMBAHIN SOFTMAX -> BUAT MODEL BARU YG LANGSUNG UBAH LOGITS JADI PROBABILITAS
probability_model(x_test[:5])
```
