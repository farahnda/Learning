```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import load_breast_cancer
cancer = load_breast_cancer()
```

```python
print(cancer['DESCR'])
```

```python
df = pd.DataFrame(cancer['data'], columns=cancer['feature_names'])
# NGUBAH ARRAY JADI TABEL
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
```

```python
X_train, X_test = train_test_split(df, test_size=0.2, random_state=42) # SPLIT DATA TEST 20%, TRAINING 80%
sc = StandardScaler() # DISCALE KARNA TIAP FITUR SATUANNYA BEDA
sc.fit(X_train)
X_train_std = sc.transform(X_train)
X_test_std = sc.transform(X_test)
```

```python
pca = PCA()

X_train_pca = pca.fit_transform(X_train_std)
# PCA COBA CARI KOMBINASI FITUR PALING BANYAK SIMPAN INFORMASI (BUAT FITUR BARU PC1 PC2 DST = GABUNGAN FITUR2 YG ADA)

exp_var_pca = pca.explained_variance_ratio_
# HITUNG BERAPA PERSEN INFORMASI YANG DIJELASKAN OLEH SETIAP PRINCIPAL COMPONENT

cum_sum_eigenvalues = np.cumsum(exp_var_pca)
# HITUNG AKUMULASI

plt.bar(range(0, len(exp_var_pca)), exp_var_pca, alpha=0.5, align='center', label='Individual Exp Variance')
# DIAGRAM BATANG ^
plt.step(range(0, len(cum_sum_eigenvalues)), cum_sum_eigenvalues, where='mid', label='Cumulative Exp Variance')
# DIAGRAM GARIS ^
plt.ylabel('Explained Variance Ratio')
plt.xlabel('PC Index')
plt.legend(loc='best')
plt.tight_layout()
plt.show()
```
