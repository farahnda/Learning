```python
import numpy as np # operasi mtk
import pandas as pd # olah data
import matplotlib.pyplot as plt # visualisaasi
import seaborn as sns # visualisasi
from sklearn import datasets # ambil dataset bawaan sklearn
from sklearn.model_selection import train_test_split # bagi data train & test
```

```python
bc = datasets.load_breast_cancer() # ambil dataset kanker payudara
x = bc.data # isi semua fitur
y = bc.target # isi label (dari target)
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42, stratify=y) # bagi 70 train 30 test
# stratify jaga biar proporsi dijaga baginya bener
```

```python
# bikin algoritma perceptron sendiri
class CustomPerceptron(object):
    def __init__(self, n_iterations=50, random_state=1, learning_rate=0.01): # constructor (init)
        self.n_iterations = n_iterations
        self.random_state = random_state
        self.learning_rate = learning_rate


    def fit(self, X, y): # MELATIH MODEL
        rgen = np.random.RandomState(self.random_state) # buat angka random
        self.coef_ = rgen.normal(loc=0.0, scale=0.01, size=1 + X.shape[1]) # buat bobot awal (misal dataset punya 2 fitur, maka size = 3, karna 2 bobot + 1 bias)
        for _ in range(self.n_iterations): # model akan liat seluruh data sebanyak 50x, 1x putaran = epoch
            for xi, expected_value in zip(X, y): # 1 1 dipelajari
                predicted_value = self.predict(xi)
                self.coef_[1:] += self.learning_rate * (expected_value - predicted_value) * xi # RUMUS UTAMA PERCEPTRON (wbaru = wlama + n(y-y^)x)
                self.coef_[0] += self.learning_rate * (expected_value - predicted_value) * 1 # UPDATE BIAS

    def activation(self, X): # HITUNG (z = w^T x + b)
        return np.dot(X, self.coef_[1:]) + self.coef_[0]

    def predict(self, X):
        output = self.activation(X)
        return np.where(output >= 0.0, 1, 0)

    def score(self, X, y):
        misclassified_data_count = 0 # HITUNG AKURASI
        for xi, target in zip(X, y):
            output = self.predict(xi)
            if (target != output):
                misclassified_data_count += 1
        total_data_count = len(X)
        self.score_ = (total_data_count - misclassified_data_count) / total_data_count
        return self.score_
```

```python
n_iterations = 100
learning_rate = 0.01
```

```python
prcptrn = CustomPerceptron() # BUAT OBJEK PERCEPTRON
prcptrn.fit(X_train, y_train) # MODEL BELAJAR
# (bobot diacak -> data 1st dipredict, if false -> data 2nd-> until its done w/ n_iteration, then save it to self.coef_)
```

```python
print(prcptrn.score(X_test, y_test))
prcptrn.score(X_train, y_train)
```
